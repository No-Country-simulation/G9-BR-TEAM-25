```python
media_resumo = df_eda.groupby("categoria_projeto")["tamanho_resumo"].mean().sort_values()
n_total = len(media_resumo)
n_destaque = 5

cores_barra = ['#D9D9D9' if i < n_total - n_destaque else '#1B7F79' for i in range(n_total)]
cores_texto = ['#999' if i < n_total - n_destaque else '#333' for i in range(n_total)]

ax = media_resumo.plot(kind="barh", color=cores_barra, figsize=(8,5))
ax.set_title("Tamanho Médio do Resumo por Categoria")
ax.set_xlabel("")
ax.set_ylabel("")

for i, label in enumerate(ax.get_yticklabels()):
    label.set_fontweight("bold" if i >= n_total - n_destaque else "normal")
    label.set_color(cores_texto[i])

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.xaxis.set_visible(False)

for i, v in enumerate(media_resumo):
    ax.text(v + 1, i, f"{v:.0f}", va="center", color=cores_texto[i])

plt.show()
```

**`Análise Textual`**

```python
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

stop_words_custom = stop_words.union({
    "study",
    "show",
    "present",
    "present",
    "result",
    "results",
    "using",
    "however",
    "method",
    "methods",
    "approach",
    "paper",
    "existing",
    "also"
   })
```

```python
from collections import Counter

texto = " ".join(df_eda["resumo_limpo"]).lower()

palavras = texto.split()

palavras_sem_stopwords = [
    p for p in palavras if p not in stop_words_custom
]

frequencia = Counter(palavras_sem_stopwords)

palavras_mais_comuns = frequencia.most_common(20)

palavras_mais_comuns
```

```python
import matplotlib.pyplot as plt
import pandas as pd

serie_palavras = pd.Series(dict(palavras_mais_comuns)).sort_values()

# Mantém só as 10 mais frequentes
top10 = serie_palavras.tail(10)

fig, ax = plt.subplots(figsize=(10, 6))

n_destaque = 5
cores = ['#D9D9D9'] * len(top10)
for i in range(1, n_destaque + 1):
    cores[-i] = '#1B7F79'

bars = ax.barh(top10.index, top10.values, color=cores)

for spine in ax.spines.values():
    spine.set_visible(False)
ax.xaxis.set_visible(False)

for i, bar in enumerate(bars):
    largura = bar.get_width()
    destaque = i >= len(top10) - n_destaque
    ax.text(largura + max(top10.values)*0.01,
            bar.get_y() + bar.get_height()/2,
            f'{int(largura)}', va='center', fontsize=9,
            color='#333' if destaque else '#999',
            fontweight='bold' if destaque else 'normal')

labels = ax.get_yticklabels()
for i, label in enumerate(labels):
    if i >= len(top10) - n_destaque:
        label.set_fontweight('bold')
        label.set_color('#333')
    else:
        label.set_color('#999')

ax.set_title("Top 10 palavras mais frequentes nos resumos dos artigos",
             fontsize=14, fontweight='bold', pad=15)
ax.set_ylabel("")
ax.tick_params(axis='y', labelsize=10, length=0)

plt.tight_layout()
plt.show()
```

Conclusões

A base de dados é composta por 5.963 artigos publicados entre 2021 e 2026. A distribuição temporal está concentrada nos anos mais recentes, com média de publicação em 2023 e mediana em 2024, indicando predominância de estudos publicados entre 2023 e 2025. A coluna id funciona apenas como identificador único de registros, sem ser utilizada para análises estatísticas.

Para conferir a distribuição inicial dos dados, observou-se que a quantidade de artigos por categoria de projetos é similar, variando entre 565 e 600 artigos por categoria.

Da mesma forma, a distribuição temporal dos artigos apresentou comportamento equilibrado, com quantidades próximas entre os anos analisados, variando de 978 a 1000 artigos por ano no período de 2021 a 2026.


A análise do tamanho do resumo dos artigos mostrou que a média de palavras é de 121, sendo que metade dos resumos apresenta até 122 palavras e metade possui mais de 122 palavras, sendo que 50% possuem entre 97 e 146 palavras. E um ponto de atenção foi o menor resumo contar com apenas 7 palavras. Isso levou a uma investigação de resumos com menos de 20 palavras, que mostrou 5 materiais com essa característica.



Para complementar essas observações foram produzidos gráficos para demonstrar: a distribuição de artigos por categoria; a evolução da quantidade de artigos ao longo dos anos; o tamanho médio dos resumos por categoria; e a distribuição do tamanho dos resumos da base.

Partindo das informações do tamanho do resumo, investigou-se as palavras mais frequentes organizadas por categoria do projeto, considerando algumas stopwords do inglês e customizando essa variável para palavras adicionais.

Foi realizado o pré-processamento textual dos resumos, incluindo a remoção de stopwords em inglês e de termos genéricos recorrentes na literatura científica. Posteriormente, utilizou-se a função Counter para identificar as palavras mais frequentes. Para facilitar a compreensão desses dados, também foi elaborado um gráfico de barras horizontais, destacando as principais 5 palavras presentes nos resumos.
